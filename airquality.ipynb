## Examining Air Quality in Indian Cities


```python
import pandas as pd 
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("data.csv", encoding='latin1')

data=pd.DataFrame(data)
```

```python
print(data.head())
print(data.info())
```

```python
average_pm2_5 = data["pm2_5"].mean()
print(average_pm2_5)
```

```python
state_highest_no2 = data.loc[data['no2'].idxmax(), 'state']
print("\nState with highest NO2 levels:", state_highest_no2)
```

```python
correlation_pm2_5_no2 = data['pm2_5'].corr(data['no2'])
print("\nCorrelation between PM2.5 and no2:", correlation_pm2_5_no2)
```

```python
# Filter top 20 states based on PM2.5 levels
top_20_states = data.nlargest(20, 'pm2_5')

# Plot the bar chart
plt.figure(figsize=(11, 6))
sns.barplot(x='state', y='pm2_5', data=top_20_states)
plt.title('Top 20 States by PM2.5 Levels')
plt.xlabel('State')
plt.ylabel('PM2.5')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(13, 9))

sns.scatterplot(x='pm2_5', y='no2', data=data, hue='state')
plt.title('PM2.5 vs. NO2 Levels')

plt.xlabel('PM2.5')
plt.ylabel('NO2')

plt.show()
```
